## def & return
1.Принимает список строк.
2.Удаляет пробелы по краям каждого элемента.
3.Приводит все строки к верхнему регистру.
4.Удаляет дубликаты.
5.Возвращает результат в виде множества (set).

```python
def normalize_items(items):
    clean_list = [item.strip().upper() for item in items]
    unique_set = set(clean_list)
    return unique_set

test_items = [
    " a-101 ",
    "A-102",
    "a-101",
    " A-103 ",
    "a-102 "
]

clean_items_set = normalize_items(test_items)

print(clean_items_set)
```
